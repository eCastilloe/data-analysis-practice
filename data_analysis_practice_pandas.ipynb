##Carga de datos

```python
import pandas as pd

df = pd.read_csv('ventas.csv')
#Imprimir la tabla
print(df)
```

##Exploración inicial

```python
#Imprimir solo las primeras 5 filas
print(df.head())

#Filtrar por columna
print(df['producto'])

#Filtrar filas por condición
guadalajara = df[df['region'] == 'Zapopan']
print(guadalajara)
```

##Análisis por región

```python
#Agrupar y sumar
por_region = df.groupby('region')['cantidad'].sum()
print(por_region)

#Varias operaciones a la vez
resumen = df.groupby('region')['cantidad'].agg(['sum','mean','count'])
print(resumen)
```

##Visualización

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(data=df, x='producto', y='cantidad')
plt.title('Cantidad total vendida por producto')
plt.show()
```

##Conclusión

En los productos disponibles, podemos observar que el film envolvente es el que mejor ventas tiene, con casi el doble que el segundo producto más vendido tanto en cantidad total como en promedio.
